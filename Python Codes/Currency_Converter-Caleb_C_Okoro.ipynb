```python
# ...existing code...
"""
Simple NGN currency converter.

Accepts an amount in Nigerian Naira (NGN) from the user and converts it
to multiple target currencies using rates defined in a dictionary.
"""

def get_ngn_amount():
    """Prompt the user for an NGN amount, validate and return it as float.

    Repeatedly asks until a non-negative numeric value is entered.
    Removes commas to allow inputs like '1,000'.
    """
    while True:
        s = input("Enter amount in Nigerian Naira (NGN): ").strip().replace(',', '')
        try:
            amt = float(s)
            if amt < 0:
                print("❌ Please enter a non-negative amount.\n")  # Added emoji and newline
                continue
            return amt
        except ValueError:
            print("❌ Invalid number, try again.\n")  # Added emoji and newline

def convert(amount_ngn, rates):
    """Convert an NGN amount to target currencies.

    Args:
        amount_ngn (float): amount in NGN to convert.
        rates (dict): mapping of currency codes to exchange rate
                      expressed as 1 NGN in target currency.

    Returns:
        dict: mapping of currency codes to converted amounts.
    """
    return {cur: amount_ngn * rate for cur, rate in rates.items()}

def display_results(amount_ngn, converted, symbols):
    """Print a formatted summary of the original and converted amounts.

    Args:
        amount_ngn (float): original amount in NGN.
        converted (dict): mapping currency code -> converted amount.
        symbols (dict): mapping currency code -> currency symbol string.
    """
    print("\nConversion results")
    print(f"NGN: ₦{amount_ngn:,.2f}")
    for cur, val in converted.items():
        sym = symbols.get(cur, '')
        print(f"{cur}: {sym}{val:,.2f}")

def main():
    """Main program flow: define rates, read input, convert and display."""
    rates = {
        "USD": 0.0026,    # 1 NGN = 0.0026 USD (US Dollar), Hence 1 USD = 384.62 NGN
        "EUR": 0.0024,    # 1 NGN = 0.0024 EUR (Euro), Hence 1 EUR = 416.67 NGN
        "GBP": 0.0021     # 1 NGN = 0.0021 GBP (British Pound), Hence 1 GBP = 476.19 NGN
    }
    symbols = {"USD": "$", "EUR": "€", "GBP": "£"}

    amount_ngn = get_ngn_amount()
    converted = convert(amount_ngn, rates)
    display_results(amount_ngn, converted, symbols)

if __name__ == "__main__":
    main()
# ...existing code...
```
